### Writing your OpenRouter code for the OpenAI model

````python
# --- Student TODO: implement only this function. Keep it minimal. ---
from openai import OpenAI

# Stores token usage of the most recent call per model (used by the playground below).
LAST_USAGE = {}

def _clean(text: str) -> str:
    """Strip whitespace and any accidental ```json fences so JSON output parses."""
    text = (text or "").strip()
    if text.startswith("```"):
        text = text.split("\n", 1)[1] if "\n" in text else ""
        if text.rstrip().endswith("```"):
            text = text.rstrip()[:-3]
    return text.strip()

def _chat(model: str, prompt: str) -> str:
    """Send one user prompt to `model` through OpenRouter and return plain text."""
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,   # modest temperature: some creativity, stable formatting
        max_tokens=200,    # cap output length to control cost/latency
    )
    LAST_USAGE[model] = getattr(response, "usage", None)
    return _clean(response.choices[0].message.content)

def call_openai(prompt: str) -> str:
    """Return the plain text output from OPENAI_MODEL through OpenRouter."""
    return _chat(OPENAI_MODEL, prompt)
````

## What we will test (assert cells)
We will call **your three functions** with three prompts and check shape/format only:

**Prompts we will use:**
```python
CREATIVE_PROMPT = "Write a 4-line poem about learning by testing code."
ANALYTICAL_PROMPT = (
    "Summarize in one sentence (<= 40 words): "
    "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
    "but adds integration overhead and testing complexity.'"
)
INSTRUCTION_PROMPT = (
    "Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
    "which is a list of EXACTLY 2 items. Each item must be an object with keys 'task' (string) "
    "and 'done' (boolean)."
)
```

**Checks:**
- Creative: output has **≥ 4 non-empty lines**.
- Analytical: output is **one sentence** and **≤ 40 words**.
- Instruction: output is **valid JSON** with exactly **2** todos, each `{ "task": str, "done": bool }`.

That’s all. Keep it simple.


````python
# ---------------- Multibot Chat Playground: prompts, models, checks ----------------
import json, re, time

CREATIVE_PROMPT = "Write a 4-line poem about learning by testing code."
ANALYTICAL_PROMPT = (
    "Summarize in one sentence (<= 40 words): "
    "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
    "but adds integration overhead and testing complexity.'"
)
INSTRUCTION_PROMPT = (
    "Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
    "which is a list of EXACTLY 2 items. Each item must be an object with keys 'task' (string) "
    "and 'done' (boolean)."
)
TASKS = {"creative": CREATIVE_PROMPT, "analytical": ANALYTICAL_PROMPT, "instruction": INSTRUCTION_PROMPT}

# Model selection: add/remove entries here to switch models in the playground.
MODELS = {
    "OpenAI (" + OPENAI_MODEL + ")": (call_openai, OPENAI_MODEL),
    "Gemini (" + GEMINI_MODEL + ")": (call_gemini, GEMINI_MODEL),
    "Claude (" + CLAUDE_MODEL + ")": (call_claude, CLAUDE_MODEL),
}

def check(task: str, out: str) -> str:
    """Return 'PASS' or 'FAIL: reason' using the autograder's criteria."""
    if task == "creative":
        n = len([l for l in out.splitlines() if l.strip()])
        return "PASS" if n >= 4 else f"FAIL: {n} lines"
    if task == "analytical":
        words = len(out.split())
        sents = [s for s in re.split(r"(?<=[.!?])\s+", out.strip()) if s]
        if not out: return "FAIL: empty"
        if words > 40: return f"FAIL: {words} words"
        if len(sents) != 1 or out.strip()[-1] not in ".!?": return "FAIL: not one sentence"
        return "PASS"
    try:
        data = json.loads(out)
        ok = (isinstance(data, dict) and set(data) == {"todos"}
              and isinstance(data["todos"], list) and len(data["todos"]) == 2
              and all(isinstance(i, dict) and set(i) == {"task", "done"}
                      and isinstance(i["task"], str) and isinstance(i["done"], bool)
                      for i in data["todos"]))
        return "PASS" if ok else "FAIL: schema"
    except Exception:
        return "FAIL: invalid JSON"
````

````python
# ---------------- Run every task through every model (same prompt each time) ----------------
results = []
for task, prompt in TASKS.items():
    print("=" * 80, f"\nTASK: {task.upper()}\nPROMPT: {prompt}\n")
    for label, (fn, model_id) in MODELS.items():
        start = time.time()
        try:
            out = fn(prompt)
            err = None
        except Exception as e:           # keep going if one provider fails
            out, err = "", f"{type(e).__name__}: {e}"
        secs = round(time.time() - start, 2)
        usage = LAST_USAGE.get(model_id)
        results.append({
            "task": task, "model": label, "seconds": secs,
            "prompt_tokens": getattr(usage, "prompt_tokens", None),
            "completion_tokens": getattr(usage, "completion_tokens", None),
            "words": len(out.split()),
            "check": err or check(task, out),
            "output": out,
        })
        print(f"--- {label}  |  {secs}s  |  {results[-1]['check']} ---\n{out or err}\n")
````

````python
# ---------------- Summary table of collected outputs ----------------
import pandas as pd
df = pd.DataFrame(results)
display(df.drop(columns="output"))
print("\nAverage latency (s) per model:")
print(df.groupby("model")["seconds"].mean().round(2))
print("\nChecks passed per model:")
print(df.assign(ok=df["check"].eq("PASS")).groupby("model")["ok"].sum())
````

## Reflection

This playground sent three identical prompts (a four-line poem, a one-sentence summary, and a strict JSON to-do list) to three models from different providers, GPT-4o-mini, Gemini 2.5 Flash, and Claude Sonnet 4.5, all through one OpenRouter endpoint. Keeping the prompt, temperature (0.3), and token cap (200) fixed meant the model was the only real variable, so the differences in the output table can be attributed to the models themselves rather than to my setup.

**Which model handled each task best, and why?**
For creative writing, Claude produced the most polished poem. Its lines had a clearer arc, from writing a test, to watching it fail, to understanding the bug, and its imagery felt less generic. GPT-4o-mini wrote a tidy, rhyming quatrain that met the brief but leaned on familiar phrases such as "lines of code" and "bugs take flight." Gemini was lively and fluent, but it sometimes added a title line or a short closing remark, which is harmless for the line-count check but shows a looser reading of "write a poem." For analytical reasoning, all three models captured both sides of the trade-off (better comparison of creativity, formatting, and speed versus more integration and testing work). GPT-4o-mini and Claude were the most disciplined about staying in one sentence well under 40 words. Claude usually phrased the tension most precisely. For instruction following, GPT-4o-mini and Claude reliably returned bare, valid JSON with exactly two todos. Gemini was the model most likely to wrap the JSON in a markdown code fence, which would break a strict `json.loads` check. That is why the helper function strips fences before returning. Overall, Claude was the best writer, GPT-4o-mini was the most predictable formatter for the price, and Gemini was fast and capable but needed the most guarding.

**How did speed, formatting, tone, and creativity vary?**
The latency column showed the smaller "mini/flash" models answering noticeably faster than the larger Claude model. Gemini Flash was often the quickest, although OpenRouter adds a small, variable routing overhead to every call. Tone differed in consistent ways. GPT-4o-mini was neutral and compact. Claude was warmer and more deliberate. Gemini was upbeat and occasionally chattier than asked. Creativity followed model size more than provider, with the larger model taking more interesting risks with metaphor. Formatting was the most important difference for engineering purposes. A poem with an extra title line still passes, but JSON with a single stray backtick fails completely, so formatting reliability is a hard requirement rather than a style preference.

**What API mechanics influenced the results?**
Several parameters mattered. Because OpenRouter exposes an OpenAI-compatible `/chat/completions` endpoint, one `OpenAI` client with a different `base_url` could reach every provider, and switching models only meant changing the `model` string. The `max_tokens=200` cap kept costs and latency down, but it is also a risk. Models that spend tokens on hidden reasoning, as some Gemini 2.5 configurations can, may leave little budget for the visible answer, so truncated or empty output is something to monitor. A low temperature of 0.3 improved format stability on the JSON and one-sentence tasks at a slight cost to poetic variety. The `usage` object returned with each response made prompt and completion tokens visible. That showed the prompts cost almost the same across models while completion lengths differed, and output tokens drive most of the cost. Model IDs themselves also turned out to be a practical concern. Older IDs such as Claude 3.5 Sonnet get retired, so model names should live in configuration variables rather than being scattered through the code.

**How would I route tasks to the best-fit model automatically?**
I would build a router in three layers. First, a lightweight classifier (keyword rules or a very cheap model) would label each request as creative, analytical, or structured output. Second, a policy table would map each label to a preferred model and a fallback, informed by exactly the kind of measurements this notebook collects: for example, Claude for creative writing, GPT-4o-mini for strict JSON, and Gemini Flash for latency-sensitive summaries. Third, a validation step would check the output against the task's contract, such as parsing the JSON, counting words, or counting lines. If validation fails, the router would retry once with a corrective message and then escalate to the fallback model. Logging latency, token usage, and pass rates for every call would let the policy table be updated from real data instead of intuition, and OpenRouter's single key and uniform response schema make this kind of routing cheap to implement.

**Strengths and limitations of using multiple providers**
The strengths are clear. No single model won every task, so combining them lets an application get the best writing, the most reliable formatting, and the lowest latency where each matters most. Multiple providers also bring resilience: if one provider fails or retires a model, the try/except in the playground shows how the others can keep the application working. The limitations are equally real. Each model has its own formatting habits, so output cleaning and validation become mandatory. Behavior can shift when a provider updates a model, so tests must be rerun regularly. Latency and cost vary per call, which complicates budgeting. A gateway such as OpenRouter reduces integration work, but it adds another dependency and some overhead. Multi-model systems pay off when they come with automated evaluation like the checks in this notebook. Without that testing layer, extra providers mostly add ways to fail.
